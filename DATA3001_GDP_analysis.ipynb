```python
gulf_mask = (
    (ds["lat"] >= 25) &
    (ds["lat"] <= 30) &
    (ds["lon"] >= -93) &
    (ds["lon"] <= -85)
)

gulf_mask
```

```python
gulf_count = gulf_mask.sum().compute()

print("North-central Gulf observations:", gulf_count.item())
```

```python
import numpy as np

obs_id = np.repeat(
    ds["id"].values,
    ds["rowsize"].values
)

print("Total observations:", len(obs_id))
print("First 10 observation IDs:", obs_id[:10])
```

```python
gulf_ids = np.unique(obs_id[gulf_mask.values])

print("Distinct trajectories entering North-central Gulf:", len(gulf_ids))
```

```python
gulf_lat = ds["lat"].values[gulf_mask.values]
gulf_lon = ds["lon"].values[gulf_mask.values]
gulf_obs_id = obs_id[gulf_mask.values]

print("Observations:", len(gulf_lat))
print("Example:")
print(gulf_lat[:5])
print(gulf_lon[:5])
print(gulf_obs_id[:5])
```

```python
lat_cell = np.floor(gulf_lat)
lon_cell = np.floor(gulf_lon)

print("Example cells:")
for i in range(5):
    print(
        "Point:", gulf_lat[i], gulf_lon[i],
        "→ Cell:", lat_cell[i], lon_cell[i]
    )
```

```python
from collections import defaultdict

cell_trajectories = defaultdict(set)

for lat_c, lon_c, drifter_id in zip(lat_cell, lon_cell, gulf_obs_id):
    cell_trajectories[(lat_c, lon_c)].add(drifter_id)

cell_counts = {
    cell: len(ids)
    for cell, ids in cell_trajectories.items()
}

cell_counts
```

```python
import matplotlib.pyplot as plt

x = [cell[1] + 0.5 for cell in cell_counts]
y = [cell[0] + 0.5 for cell in cell_counts]
counts = list(cell_counts.values())

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    x,
    y,
    c=counts,
    s=600,
    marker="s"
)

plt.colorbar(scatter, label="Distinct trajectories")

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Distinct trajectories by 1° cell — North-central Gulf")

plt.xlim(-93, -85)
plt.ylim(25, 30)

plt.grid(alpha=0.2)
plt.show()
```

```python
traj_mask = np.isin(ds["id"].values, gulf_ids)

print("Selected trajectories:", traj_mask.sum())
```

```python
rowsize = ds["rowsize"].values

starts = np.concatenate(([0], np.cumsum(rowsize)[:-1]))
ends = np.cumsum(rowsize)

print("First trajectory:")
print("ID:", ds["id"].values[0])
print("Observation range:", starts[0], "to", ends[0] - 1)
print("Number of observations:", rowsize[0])
```

```python
valid_30d = []

for k in np.where(traj_mask)[0]:

    # 这条 trajectory 在总数据中的范围
    s = starts[k]
    e = ends[k]

    lat = ds["lat"].values[s:e]
    lon = ds["lon"].values[s:e]
    time = ds["time"].values[s:e]

    # 找这条 trajectory 哪些点进入了 North-central Gulf
    inside = (
        (lat >= 25) & (lat <= 30) &
        (lon >= -93) & (lon <= -85)
    )

    # 第一次进入 Gulf = t0
    first = np.where(inside)[0][0]

    t0 = time[first]
    target = t0 + np.timedelta64(30, "D")

    # 找距离30天最近的 observation
    diff = np.abs(time - target)
    nearest = np.argmin(diff)

    # Google Doc 当前规则：30 days ± 3 hours
    if diff[nearest] <= np.timedelta64(3, "h"):
        valid_30d.append({
            "id": ds["id"].values[k],
            "t0": t0,
            "origin_lat": lat[first],
            "origin_lon": lon[first],
            "dest_time": time[nearest],
            "dest_lat": lat[nearest],
            "dest_lon": lon[nearest]
        })

print("Trajectories entering Gulf:", len(gulf_ids))
print("Valid 30-day trajectories:", len(valid_30d))
```

```python
import pandas as pd

df30 = pd.DataFrame(valid_30d)

df30.head()
```

```python
plt.figure(figsize=(10, 7))

plt.scatter(
    df30["dest_lon"],
    df30["dest_lat"],
    alpha=0.6
)

# 把 North-central Gulf source region 画出来
plt.plot(
    [-93, -85, -85, -93, -93],
    [25, 25, 30, 30, 25],
    linewidth=2
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Destinations 30 Days After First Entry into North-central Gulf")

plt.grid(alpha=0.2)
plt.show()
```

```python
# Put each 30-day destination into a 1° × 1° cell
df30["dest_lon_cell"] = np.floor(df30["dest_lon"])
df30["dest_lat_cell"] = np.floor(df30["dest_lat"])

# Count destinations in each cell
prob = (
    df30.groupby(["dest_lon_cell", "dest_lat_cell"])
        .size()
        .reset_index(name="count")
)

# Convert counts to empirical probabilities
prob["probability"] = prob["count"] / len(df30)

prob = prob.sort_values("probability", ascending=False)

prob.head(10)
```

```python
plt.figure(figsize=(11, 7))

sc = plt.scatter(
    prob["dest_lon_cell"] + 0.5,
    prob["dest_lat_cell"] + 0.5,
    c=prob["probability"],
    s=500,
    marker="s"
)

plt.colorbar(sc, label="Empirical destination probability")

# Source region
plt.plot(
    [-93, -85, -85, -93, -93],
    [25, 25, 30, 30, 25],
    linewidth=2
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("30-Day Destination Probability — North-central Gulf")

plt.grid(alpha=0.2)
plt.show()
```
